```python
import pandas as pd
```

```python
wygo = pd.read_csv("Spieler_Statistik_25_26.csv")
```

```python
wygo['Plus'] = wygo["Plus"].replace('-', 0)
wygo['Minus'] = wygo["Minus"].replace('-', 0)
wygo['Minus'] = wygo["Minus"].astype(int)
wygo['Plus'] = wygo["Plus"].astype(int)
```

```python
wygo1 = wygo.groupby('Name')[['Plus', 'Minus']].sum()
wygo1['Plus/Minus'] = wygo1['Plus'] - wygo1['Minus']
wygo1.sort_values(by='Plus/Minus', ascending=False)
```
